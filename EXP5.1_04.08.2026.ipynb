```python
import pandas as pd
weight  = [40,43,45,48,49,52,56,55,59,60,63,65,62,67,66,68,70,75,77,73,80,85]
bins = [40,50,60,70,80]
category = pd.cut(weight,bins)
print(category)
```

```python
print(pd.value_counts(category))
```

```python
bin_name = ['under weight','normal weight','over weight','obesity']
print(pd.cut(weight,bins,labels=bin_name))
```
