```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
df = pd.read_csv('/content/fraudTest.csv')
columns_to_drop = []
if 'Time' in df.columns:
  columns_to_drop.append('Time')
if 'Class' in df.columns:
  columns_to_drop.append('Class')
features_to_normalize = df.drop(columns=columns_to_drop, errors='ignore').select_dtypes(include=np.number).columns
data_to_normalize = df[features_to_normalize]
min_max_scaler = MinMaxScaler()
df_min_max_scaled = pd.DataFrame(min_max_scaler.fit_transform(data_to_normalize), columns=features_to_normalize)

print("DataFrame after Min-Max Scaling (first 5 rows):")
print(df_min_max_scaled.head())
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
standard_scaler = StandardScaler()
df_standard_scaled = pd.DataFrame(standard_scaler.fit_transform(data_to_normalize), columns=features_to_normalize)
print("DataFrame after Standardization (first 5 rows):")
display(df_standard_scaled.head())
```
