```python
import numpy as np
import pandas as pd

np.random.seed(42)
skewed_data = np.random.lognormal(mean=3, sigma=1, size=1000)
df = pd.DataFrame({'Original_Feature': skewed_data})
print("Skewness BEFORE Log Transform:", df['Original_Feature'].skew())

df['Log_Transformed_Feature'] = np.log1p(df['Original_Feature'])

print("Skewness AFTER Log Transform:", df['Log_Transformed_Feature'].skew())

# Print sample rows to see the compression effect
print("\nSample Data Comparison:")
print(df.head())
```

```python
import numpy as np
import pandas as pd
# Sample transaction dataset in list form
transactions = [
    {"Transaction_ID": 101, "Category": "Groceries", "Amount": 12.50, "Risk_Score": 0.12},
    {"Transaction_ID": 102, "Category": "Coffee", "Amount": 4.75, "Risk_Score": 0.05},
    {"Transaction_ID": 103, "Category": "Electronics", "Amount": 1250.00, "Risk_Score": 0.85}, # Outlier/Skew creator
    {"Transaction_ID": 104, "Category": "Groceries", "Amount": 45.20, "Risk_Score": 0.18},
    {"Transaction_ID": 105, "Category": "Gas", "Amount": 35.00, "Risk_Score": 0.11},
    {"Transaction_ID": 106, "Category": "Retail", "Amount": 89.99, "Risk_Score": 0.22},
    {"Transaction_ID": 107, "Category": "Fast Food", "Amount": 8.40, "Risk_Score": 0.09},
    {"Transaction_ID": 108, "Category": "Luxury Watch", "Amount": 4500.00, "Risk_Score": 0.92}, # Outlier/Skew creator
    {"Transaction_ID": 109, "Category": "Gas", "Amount": 28.50, "Risk_Score": 0.10},
    {"Transaction_ID": 110, "Category": "Movies", "Amount": 15.00, "Risk_Score": 0.14}
]


# 1. Load the list into a pandas DataFrame
df = pd.DataFrame(transactions)

# 2. Print initial skewness (Expect a high positive number due to the 1250 and 4500 values)
print("Original Amount Skewness:", df['Amount'].skew())

# 3. Apply the log transformation (using log1p to safely handle potential 0 values)
df['Amount_Log'] = np.log1p(df['Amount'])

# 4. Print post-transformation skewness (The value will drop significantly closer to 0)
print("Log Transformed Amount Skewness:", df['Amount_Log'].skew())

# View the final dataset side-by-side
print("\nTransformation Impact:")
print(df[['Amount', 'Amount_Log']])
```

```python
import pandas as pd
from sklearn.preprocessing import PowerTransformer

# 1. Your transaction dataset in list form
transactions = [
    {"Transaction_ID": 101, "Category": "Groceries", "Amount": 12.50, "Risk_Score": 0.12},
    {"Transaction_ID": 102, "Category": "Coffee", "Amount": 4.75, "Risk_Score": 0.05},
    {"Transaction_ID": 103, "Category": "Electronics", "Amount": 1250.00, "Risk_Score": 0.85},
    {"Transaction_ID": 104, "Category": "Groceries", "Amount": 45.20, "Risk_Score": 0.18},
    {"Transaction_ID": 105, "Category": "Gas", "Amount": 35.00, "Risk_Score": 0.11},
    {"Transaction_ID": 106, "Category": "Retail", "Amount": 89.99, "Risk_Score": 0.22},
    {"Transaction_ID": 107, "Category": "Fast Food", "Amount": 8.40, "Risk_Score": 0.09},
    {"Transaction_ID": 108, "Category": "Luxury Watch", "Amount": 4500.00, "Risk_Score": 0.92},
    {"Transaction_ID": 109, "Category": "Gas", "Amount": 28.50, "Risk_Score": 0.10},
    {"Transaction_ID": 110, "Category": "Movies", "Amount": 15.00, "Risk_Score": 0.14}
]

# 2. Convert the list to a DataFrame
df = pd.DataFrame(transactions)

# 3. Check the skewness before transformation
print("Original Amount Skewness:", df['Amount'].skew())

# 4. Initialize PowerTransformer (Box-Cox works perfectly since all amounts are > 0)
pt = PowerTransformer(method='box-cox', standardize=True)

# 5. Fit and transform the 'Amount' column
df['Amount_Power'] = pt.fit_transform(df[['Amount']])

# 6. Check the skewness after transformation
print("Power Transformed Amount Skewness:", df['Amount_Power'].skew())

# 7. View the final side-by-side comparison
print("\nTransformation Results:")
print(df[['Amount', 'Amount_Power']])
```
